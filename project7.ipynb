```python
#define the robert kernels
kernel_gx=np.array([[1,0],[0,-1]])
kernel_gy=np.array([[0,1],[-1,0]])
```

```python
kernel_gx
```

```python
kernel_gy
```

```python
#covaulation of two arrays
gradient_x=cv2.filter2D(img1_gray,cv2.CV_32F,kernel_gx)
gradient_y=cv2.filter2D(img1_gray,cv2.CV_32F,kernel_gy)
```

```python
gradient_x2=cv2.filter2D(img2_gray,cv2.CV_32F,kernel_gx)
gradient_y2=cv2.filter2D(img2_gray,cv2.CV_32F,kernel_gy)
```

```python
gradient_x.dtype,gradient_x2.dtype
```

```python
plt.subplot(1,2,1)
plt.imshow(gradient_x,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(gradient_x2,cmap='gray')
```

```python
# calulate the graident magnitude
gradient_mag=np.sqrt(gradient_x**2+gradient_y**2)
gradient_mag2=np.sqrt(gradient_x2**2+gradient_y2**2)
```

```python
plt.subplot(1,2,1)
plt.imshow(gradient_mag,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(gradient_mag2,cmap='gray')
```

```python
threshold_value=20
threshold_value2=110
edges=(gradient_mag>threshold_value)
edges2=(gradient_mag2>threshold_value2)
```

```python
plt.subplot(1,2,1)
plt.imshow(edges,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(edges2,cmap='gray')
```

```python
#5. Prewitt Edge Detector — Kernel Size = 3
kernel_gx = np.array([[1,0,-1],
                      [1,0,-1],
                      [1,0,-1]])

kernel_gy = np.array([[1,1,1],
                      [0,0,0],
                      [-1,-1,-1]])
```

```python
gradient_x = cv2.filter2D(gray, cv2.CV_32F, kernel_gx)
gradient_y = cv2.filter2D(gray, cv2.CV_32F, kernel_gy)

gradient_x2 = cv2.filter2D(gray2, cv2.CV_32F, kernel_gx)
gradient_y2 = cv2.filter2D(gray2, cv2.CV_32F, kernel_gy)
```

```python
gradient_mag = np.sqrt(gradient_x**2 + gradient_y**2)
gradient_mag2 = np.sqrt(gradient_x2**2 + gradient_y2**2)
```

```python
plt.subplot(1,2,1)
plt.imshow(gradient_mag, cmap='gray')

plt.subplot(1,2,2)
plt.imshow(gradient_mag2, cmap='gray')
```

```python
# Sobel Edge Detection — Kernel Size = 3
sobel_x = cv2.Sobel(gray, cv2.CV_32F, dx=1, dy=0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_32F, dx=0, dy=1, ksize=3)

sobel_x2 = cv2.Sobel(gray2, cv2.CV_32F, dx=1, dy=0, ksize=3)
sobel_y2 = cv2.Sobel(gray2, cv2.CV_32F, dx=0, dy=1, ksize=3)
```

```python
mag_sobel = cv2.magnitude(sobel_x, sobel_y)
mag_sobel2 = cv2.magnitude(sobel_x2, sobel_y2)
```

```python
plt.subplot(1,2,1)
plt.imshow(mag_sobel, cmap='gray')

plt.subplot(1,2,2)
plt.imshow(mag_sobel2, cmap='gray')
```
