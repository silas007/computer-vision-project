BLENDING & PASTING IMAGES

```python
img1 = cv2.imread(r"C:\Users\cl601_06\Downloads\Dog.jpg")
img2 = cv2.imread(r"C:\Users\cl601_06\Downloads\Sample.jpg")
```

```python
print(img1.shape)
print(img2.shape)
```

```python
#resizing
img1 = cv2.resize(img1, (1500,1500))
img2 = cv2.resize(img2, (1500,1500))
```

```python
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
plt.imshow(img1)
```

```python
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
plt.imshow(img2)
```

```python
blended = cv2.addWeighted(src1=img1, alpha=0.5,
                          src2=img2, beta=0.3,
                          gamma=50)
```

```python
plt.imshow(blended)
```

```python
blended1 = cv2.addWeighted(src1=img1, alpha=0.5,
                          src2=img2, beta=0.5,
                          gamma=-50)
```

```python
plt.imshow(blended1)
```
