Aim: To perform and analyze edge detection using Scharr, Canny and Laplacian operators on image data using OpenCV

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img1 = cv2.imread(r"C:\Users\ADMIN\Downloads\edge_image_detection (1).png")
img2 = cv2.imread(r"C:\Users\ADMIN\Downloads\edge_image_task (1).png")
```

```python
plt.imshow(img1)
```

```python
plt.imshow(img2)
```

```python
RGB_img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
plt.imshow(RGB_img1)
```

```python
RGB_img2= cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
plt.imshow(RGB_img2)
```

```python
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
plt.imshow(gray1, cmap='gray')
```

```python
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
plt.imshow(gray2, cmap='gray')
```

```python
scharr_x1 = cv2.Scharr(gray1, cv2.CV_64F, 1, 0)
scharr_y1 = cv2.Scharr(gray1, cv2.CV_64F, 0, 1)
```

```python
gradient_magnitude = cv2.magnitude(scharr_x1, scharr_y1)
plt.imshow(gradient_magnitude, cmap='gray')
```

```python
scharr_x2 = cv2.Scharr(gray2, cv2.CV_64F, 1, 0)
scharr_y2 = cv2.Scharr(gray2, cv2.CV_64F, 0, 1)
```

```python
gradient_magnitude2 = cv2.magnitude(scharr_x2, scharr_y2)
plt.imshow(gradient_magnitude2, cmap='gray')
```

```python
canny1 = cv2.Canny(gray1, 100, 200)
canny2 = cv2.Canny(gray2, 100, 200)
```

```python
# Canny
plt.figure(figsize=(10,10))
plt.subplot(1, 2, 1)
plt.imshow(canny1, cmap="gray")
plt.title("Edge Image Detection - Canny")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(canny2, cmap="gray")
plt.title("Edge Image Task - Canny")
plt.axis("off")
```

```python
laplacian1 = cv2.Laplacian(gray1, cv2.CV_64F)
laplacian2 = cv2.Laplacian(gray2, cv2.CV_64F)
```

```python
laplacian1 = cv2.convertScaleAbs(laplacian1)
laplacian2 = cv2.convertScaleAbs(laplacian2)
```

```python
# Laplacian
plt.figure(figsize=(10,10))
plt.subplot(1, 2, 1 )
plt.imshow(laplacian1, cmap="gray")
plt.title("Edge Image Detection - Laplacian")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(laplacian2, cmap="gray")
plt.title("Edge Image Task - Laplacian")
plt.axis("off")
```

```python
# Laplacian threshold
_, laplacian_threshold1 = cv2.threshold(
    laplacian1, 50, 255, cv2.THRESH_BINARY
)

_, laplacian_threshold2 = cv2.threshold(
    laplacian2, 50, 255, cv2.THRESH_BINARY
)
```

```python
# Laplacian Threshold
plt.figure(figsize=(10, 10))
plt.subplot(1, 2, 1)
plt.imshow(laplacian_threshold1, cmap="gray")
plt.title("Edge Image Detection - Laplacian Threshold")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(laplacian_threshold2, cmap="gray")
plt.title("Edge Image Task - Laplacian Threshold")
plt.axis("off")
```
