Import 'horses.png' and 'horse_face.png' image. Consider 'horse_face.png' as a template of 'horses.png' image.

Apply template matching using Normalized Cross-Correlation (cv2.TM_CCOEFF_NORMED) method.

Apply template matching using Sum of Squared Differences (cv2.TM_SQDIFF) method.

```python
import cv2
import tensorflow
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
```

```python
img1 = cv2.imread(r"C:\Users\ADMIN\Documents\horses.jpeg")
```

```python
plt.imshow(img1)
plt.show()
```

```python
horses = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(horses)
plt.show()
```

```python
img2 = cv2.imread(r"C:\Users\ADMIN\Documents\horse_face.jpeg")
```

```python
plt.imshow(img2)
```

```python
horse_face = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(horse_face)
```

```python
horses.shape
```

```python
horse_face.shape
```

```python
template_height,template_width,channels = horse_face.shape
```

```python
result = cv2.matchTemplate(horses,horse_face,cv2.TM_CCOEFF_NORMED)
```

```python
plt.imshow(result)
```

```python
min_val,max_val,min_loc,max_loc = cv2.minMaxLoc(result)
```

```python
print(min_val)
```

```python
max_val
```

```python
min_loc
```

```python
top_left = max_loc
```

```python
bottom_right = (top_left[0]+template_width,top_left[1]+template_height)
bottom_right
```

```python
horses_detected_template = horses.copy()
cv2.rectangle(horses_detected_template, top_left, bottom_right, (225,255,0), 25)
```

```python
plt.imshow(horses_detected_template)
plt.show()
```

```python
result2 = cv2.matchTemplate(horses,horse_face,cv2.TM_SQDIFF)
```

```python
min_val2,max_val2,min_loc2,max_loc2 = cv2.minMaxLoc(result2)
```

```python
plt.imshow(result2)
plt.show()
```

```python
top_left2 = min_loc2
```

```python
bottom_right2 = (top_left2[0]+template_width,top_left2[1]+template_height)
bottom_right2
```
