```python
img = Image.open(r"C:\Users\ADMIN\Desktop\Flower.jpg")
img = np.array(img)
```

```python
# Blue Channel
blue = img.copy()
blue[:, :, 0] = 0
blue[:, :, 1] = 0

plt.imshow(blue)
plt.title("Blue Channel")

plt.show()
```

```python
#green Channel
green = img.copy()
green[:, :, 0] = 0
green[:, :, 2] = 0

plt.imshow(green)
plt.title("Green Channel")

plt.show()
```

```python
#red Channel
red = img.copy()    
red[:, :, 1] = 0
red[:, :, 2] = 0

plt.imshow(red)
plt.title("Red Channel")

plt.show()
```

```python
img_resize = cv2.resize(img, (500, 300))
img_resize
```

```python
plt.imshow(img_resize)
```

```python
resize_img = cv2.resize(img, None, fx=0.4, fy=0.6)

plt.imshow(cv2.cvtColor(resize_img, cv2.COLOR_BGR2RGB))

plt.show()
```

```python
img = np.zeros((500, 550, 3), dtype=np.uint8)
img[:] = (0, 0, 255)

cv2.rectangle(img, (350, 300), (550, 500), (255, 255, 255), -1)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.show()
```

```python
img = np.zeros((500, 550, 3), dtype=np.uint8)
img[:] = (0, 0, 255)

cv2.line(img, (0, 0), (549, 499), (255, 255, 255), 3)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.show()
```

```python
img = np.zeros((500, 550, 3), dtype=np.uint8)
img[:]=(0,0,255)

cv2.putText(img,
            "hello",
            (50, 250),
            cv2.FONT_HERSHEY_SIMPLEX,
            5,
            (255, 255, 255),
            3)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

```python
img = np.zeros((500, 550, 3), dtype = np.uint8)
img[:] = (0, 0, 255)

center = (275, 250)

cv2.circle(img, center, 40, (255, 0, 0), -1)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

```python
pink = np.zeros((500, 550, 3), dtype=np.uint8)
pink[:] = (203, 192, 255)

cv2.ellipse(pink,
            (275, 250),
            (120, 120),
            0,
            20,
            70,
            (255, 255, 255),
            3)

plt.imshow(cv2.cvtColor(pink, cv2.COLOR_BGR2RGB))

plt.show()
```

```python
img = np.zeros((550, 550, 3), dtype=np.uint8)
img[:] = (0, 0, 255)

cv2.ellipse(img,
            (275, 275),
            (300, 200),
            0,
            0,
            360,
            (255, 255, 255),
            3)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.show()
```
