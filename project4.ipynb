```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread(r"C:\Users\ADMIN\Downloads\Fruits.png")
plt.imshow(img)
```

```python
img_hsv = cv2.cvtColor(img,cv2.COLOR_RGB2HSV)
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img_rgb)
```

```python
plt.imshow(img_hsv)
```

```python
r,g,b = cv2.split(img_rgb)
```

```python
pixels = img_rgb.reshape(-1,3)
```

```python
pixel_colors = pixels / 255.0
fig = plt.figure(figsize=(5,5),dpi = 150)
ax = plt.axes(projection = '3d')
ax.scatter(r,g,b,color = pixel_colors, marker = '.')

ax.set_xlabel('Red')
ax.set_ylabel('Green')
ax.set_zlabel('Blue')

plt.tight_layout()
plt.show()
```

```python
pixels_hsv = img_hsv.reshape(-1,3)
h,s,v = cv2.split(img_hsv)
```

```python
pixel_colors_hsv = pixels_hsv / 255.0
fig = plt.figure(figsize=(5,5),dpi = 150)
ax = plt.axes(projection = '3d')
ax.scatter(h,s,v,color = pixel_colors, marker = '.')

ax.set_xlabel('Hue')
ax.set_ylabel('Saturation')
ax.set_zlabel('Value')

plt.tight_layout()
plt.show()
```

```python
low_thresh_orange = (45,170,180)
high_thresh_orange = (120,255,255)
```

```python
orange_mask = cv2.inRange(img_hsv, low_thresh_orange, high_thresh_orange)
plt.imshow(orange_mask, cmap= 'gray')
```

```python
orange_mask_result = cv2.bitwise_and(img_rgb, img_rgb, mask= orange_mask)
plt.imshow(orange_mask_result)
```

```python
low_thresh_green = (0,60,50)
high_thresh_green = (120,255,210)
```

```python
green_mask = cv2.inRange(img_hsv, low_thresh_green, high_thresh_green)
plt.imshow(green_mask, cmap= 'gray')
```

```python
green_mask_result = cv2.bitwise_and(img_rgb, img_rgb, mask= green_mask)
plt.imshow(green_mask_result)
```

```python
low_thresh_w = (0,50,0)
high_thresh_w = (175,150,255)
```

```python
white_mask = cv2.inRange(img_hsv, low_thresh_w, high_thresh_w)
plt.imshow(white_mask, cmap= 'gray')
```

```python
white_mask_result = cv2.bitwise_and(img_rgb,img_rgb,mask=white_mask)
plt.imshow(white_mask_result)
```

```python
final = cv2.bitwise_or(orange_mask_result,green_mask_result)
f_final = cv2.bitwise_or(final,white_mask_result)
plt.imshow(final)
```

```python
kernel = np.ones((5,5), np.uint8)
morph_final = cv2.morphologyEx(f_final,cv2.MORPH_CLOSE,kernel=kernel)
plt.imshow(morph_final)
```

```python
plt.figure(figsize=(10, 10))
plt.subplot(1, 2, 1)
plt.imshow(green_mask_result, cmap="gray")

plt.subplot(1, 2, 2)
plt.imshow(final, cmap="gray")
```

```python
img2 = cv2.imread(r"C:\Users\ADMIN\Downloads\Thresh_segmentation_task.png")
plt.imshow(img2)
```

```python
# Convert BGR to RGB
img2_rgb = cv2.cvtColor(
    img2,
    cv2.COLOR_BGR2RGB
)
plt.imshow(img2_rgb)
```

```python
# Convert BGR to HSV
img2_hsv = cv2.cvtColor(
    img2,
    cv2.COLOR_BGR2HSV
)
plt.imshow(img2_hsv)
```

```python
low_thresh_ball = (0, 30, 30)
high_thresh_ball = (179, 255, 255)
```

```python
ball_mask = cv2.inRange(
    img2_hsv,
    low_thresh_ball,
    high_thresh_ball
)
```

```python
plt.imshow(ball_mask, cmap="gray")
plt.title("Ball Mask")
```

```python
ball_mask_result = cv2.bitwise_and(
    img2_rgb,
    img2_rgb,
    mask=ball_mask
)

plt.imshow(ball_mask_result)
plt.title("Segmented Balls")
```

```python
kernel = np.ones((5,5), np.uint8)

morph_ball = cv2.morphologyEx(
    ball_mask_result,
    cv2.MORPH_CLOSE,
    kernel
)

plt.imshow(morph_ball)
```

```python
plt.figure(figsize=(10, 10))
plt.subplot(1, 2, 1)
plt.imshow(ball_mask_result)
plt.title("Segmented Balls")
```
