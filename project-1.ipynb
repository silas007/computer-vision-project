##**Part-2 Blending and Pasting Images**

```python
# Import Two images ('Dog.jpg' and 'CV_Text.png' )

img1 = cv2.imread( r"C:\Users\ADMIN\Desktop\EXP-2 Image.jpg")
img2 = cv2.imread(r"C:\Users\ADMIN\Desktop\CV_text.png")
```

```python
print(img1.shape)
print(img2.shape)
```

```python
plt.imshow(img1)
```

```python
plt.imshow(img2)
```

```python
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img1)
```

```python
plt.imshow(img2)
```

##**Resizing the Images**

```python
img1 =cv2.resize(img1,(1500,1500))
img2 =cv2.resize(img2,(1500,1500))
```

```python
plt.imshow(img1)
```

```python
plt.imshow(img2)
```

```python
blended = cv2.addWeighted(src1=img1,alpha=0.5,
                          src2=img2,beta=0.3,
                          gamma=50)
```

```python
plt.imshow(blended)
```

```python
blended = cv2.addWeighted(src1=img1,alpha=0.5,src2=img2,beta=0.3,gamma=-50)
```

```python
plt.imshow(blended)
```
